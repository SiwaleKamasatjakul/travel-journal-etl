```python
import dlt
from pyspark.sql.functions import *
```

### Streaming Table 

```python
# Expectations 
rules = {
    "valid_id": "id IS NOT NULL",
    "valid_activity_code": "activity_code IS NOT NULL",
    "valid_sequence": "created_at IS NOT NULL",
}
```

```python
@dlt.table(name="DimActivityCode_Stage")

@dlt.expect_all_or_drop(rules)
# This function will do above this source
def DimActivityCode_Stage():
    df = spark.readStream.table("travel_journal_catalog.silver.activity_code")
    return df
```

```python

dlt.create_streaming_table("dim_activity_code")

dlt.apply_changes(
  target = "dim_activity_code",
  source = "DimActivityCode_Stage",
  keys = ["id"],
  sequence_by = "created_at",
  stored_as_scd_type = "1", # optional
  except_column_list=["flag"]

)
```
